<h2 style="font-size: 25px; font-weight: bold; color: #fb6227;">
  9. Libraries & Documents
</h2>

Libraries are containers to store documents and are shared across [Studio](https://console.mistral.ai/build/libraries) and [Vibe](https://chat.mistral.ai/libraries).

<img src="images/libraries.png" width="50%" height="50%">

There are 2 library types:
- **Global libraries**: you can access them from anywhere
- **Project libraries**: they are automatically created when you upload files in a project and are linked to this project


You can find the [documentation here](https://docs.mistral.ai/studio-api/knowledge-rag/libraries)

````python
from mistralai.client import Mistral
from dotenv import load_dotenv
import os

load_dotenv()
mistral = Mistral(api_key=os.environ["MISTRAL_API_KEY"])
````

<h3 style="font-size: 20px; font-weight: bold; color: #ff8f1e;">
  9.1 Libraries
</h3>

````python
# List all existing libraries
libraries = mistral.beta.libraries.list()
libraries
````

````python
for lib in libraries.data:
    print(f"Library ID: {lib.id}")
    print(f"Library Name: {lib.name}")
    print(f"Documents Count: {lib.nb_documents}")
    print(f"Description: {lib.description}\n")
````

````python
# Delete a library and its content
mistral.beta.libraries.delete(library_id="019e9761-6a1f-7402-98c6-91df95317630")
````

````python
# Create a new library
img_library = mistral.beta.libraries.create(
    name="Images for tutorials",
    description="A library to store images used in tutorial notebooks"
)

img_library
````

````python
img_library.id
````

````python
# List files from my 'images' folder
from pathlib import Path
files = [file for file in Path("images").glob("*.png")] # keep only PNG files are SVG are currently not supported
files
````

````python
from mistralai.client.models import File

# Upload files one by one
for file in files:

    with open(file, "rb") as file_content:
        uploaded_doc = mistral.beta.libraries.documents.upload(
            library_id=img_library.id,
            file=File(fileName=file.name, content=file_content)
        )
````

````python
# List files to verify the upload
response = mistral.beta.libraries.documents.list(
    library_id=img_library.id
)

for doc in response.data:
    print(f"ID: {doc.id}, Name: {doc.name}")
````

````python
# Inspect the first doc
response = mistral.beta.libraries.documents.list(library_id=img_library.id)
doc = response.data[0]
````

````python
doc.id
````

````python
doc
````

````python
# A summary is automatically generated for each file!
doc.summary
````

````python
# Find a text document in another library
doc_id = mistral.beta.libraries.documents.list(
    library_id="019e93b2-8cd4-76a4-bded-ebe721d359a5"
).data[0].id
````

````python
# Get its text content
content = mistral.beta.libraries.documents.text_content(
    library_id="019e93b2-8cd4-76a4-bded-ebe721d359a5",
    document_id=doc_id
)

# Display the first 1000 characters
from IPython.display import Markdown
Markdown(content.text[:1000])


# Unfortunately, at the time of writing we cannot access the raw bytes (file content) and therefore can't download documents
````
