The best way to learn about skills is to check the [documentation](https://agentskills.io/home) of this standard. Here's a short extract:

At its core, a skill is a folder containing a SKILL.md file. This file includes metadata (name and description, at minimum) and instructions that tell an agent how to perform a specific task. Skills can also bundle scripts, reference materials, templates, and other resources.

```text
my-skill/
├── SKILL.md          # Required: metadata + instructions
├── scripts/          # Optional: executable code
├── references/       # Optional: documentation
├── assets/           # Optional: templates, resources
└── ...               # Any additional files or directories
```

### **14.2 Get Skill**

````python
# the get method returns all fields
mistral.beta.skills.get(
    skill_id="019e7a57-1efd-7362-a177-36b4e4b9ee9c",
    alias="main" # optional
)
````

````python
# So you can use 'list' to find a skill ID if you know its name and then 'get' to load it
skill_name = "python-pep8-review"
skills = mistral.beta.skills.list().result.data

skill_id = None
for skill in skills:
    if skill.name == skill_name:
        skill_id = skill.id
        skill = mistral.beta.skills.get(skill_id=skill_id)
        print(f"Found skill ID for {skill_name}: {skill_id}")
        print(skill)
        break

if not skill_id:
    raise ValueError(f"Could not find skill {skill_name}")

````

### **14.3 Create Skill**

````python
from mistralai.client.models import SkillDefinition

new_skill = mistral.beta.skills.create(
    name="tutorial-skill",
    definition=SkillDefinition(
        description="a dummy skill for tutorial",
        body="Use this skill when..."
    ), 
    aliases=["dev"]
)

new_skill
````

### **14.4 Delete Skill**

````python
mistral.beta.skills.delete(
    skill_id=new_skill.id
)
````

### **14.5 To Go Further: Recreate your Studio Skills in Vibe CLI (or VS Code Extension)**

Check here my solution to download your skills with the Python SDK and recreate the files in your `~/.vibe/skills` folder (=global skills)

https://www.reddit.com/r/MistralAI/comments/1urtf75/ai_studio_version_control_for_prompts_skills_in/

You could also create a skill for the other way-around (from Vibe CLI to Studio)

````python
from mistralai.client import Mistral
from mistralai.client.models import Skill
from dotenv import load_dotenv
import os
import sys
import base64

load_dotenv()
mistral = Mistral(api_key=os.environ["MISTRAL_API_KEY"])

SKILLS_DIR = os.path.expanduser("~/.vibe/skills")

def download_skill(skill_name: str) -> Skill:
    response = mistral.beta.skills.list()
    skills = [skill for skill in response.result.data]

    skill_id = None
    for skill in skills:
        if skill.name == skill_name:
            skill_id = skill.id
            break
    
    if not skill_id:
        raise ValueError(f"Could not find skill {skill_name}")
    
    skill = mistral.beta.skills.get(
        skill_id=skill_id
    )

    return skill

def create_skill_files(skill: Skill):
    skill_dir = os.path.join(SKILLS_DIR, skill.name)
    
    definition = skill.definition
    
    os.makedirs(skill_dir, exist_ok=True)
    
    skill_md_content = f"""---
name: {skill.name}
description: {definition.description}
---

    {definition.body}
    """
    
    with open(os.path.join(skill_dir, "SKILL.md"), "w") as f:
        f.write(skill_md_content)
    
    if definition.assets:
        for filepath, raw_content in definition.assets.items():
            full_path = os.path.join(skill_dir, filepath)
            os.makedirs(os.path.dirname(full_path), exist_ok=True)
                
            with open(full_path, "w") as f:
                content = base64.b64decode(raw_content.raw_content).decode('utf-8')
                f.write(content)
            
            if raw_content.is_executable:
                os.chmod(full_path, 0o755)
    
    print(f"Skill '{skill.name}' created at {skill_dir}")

if __name__ == "__main__":
    if len(sys.argv) < 2:
        print("Usage: python transfer_skill.py <skill_name>")
        sys.exit(1)
    skill_name = sys.argv[1]
    skill = download_skill(skill_name)
    create_skill_files(skill)
````
