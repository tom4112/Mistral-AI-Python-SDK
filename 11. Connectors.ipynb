<h3 style="font-size: 20px; font-weight: bold; color: #ff8f1e;">
  11.2 Use connectors with LangChain agents
</h3>

```python
from langchain.tools import tool
from langchain.agents import create_agent
from langchain_mistralai import ChatMistralAI
from mistralai.client.models import ConnectorToolCallResponse

@tool
def search_langchain_documentation(query: str)->ConnectorToolCallResponse:
    """
    Search across the Docs by LangChain knowledge base to find relevant information, 
    code examples, API references, and guides. 
    Use this tool when you need to answer questions about Docs by LangChain, 
    find specific documentation, understand how features work, or locate implementation details. 
    """
    response = mistral.beta.connectors.call_tool(
        connector_id_or_name="langchain_documentation",
        tool_name="search_docs_by_lang_chain",
        arguments={"query": query}
    )

    return response

toolbox=[search_langchain_documentation]
llm = ChatMistralAI(model="mistral-small-latest")

agent = create_agent(
    model=llm.bind_tools(toolbox),
    tools=toolbox,
    system_prompt="Use your tools to respond to questions"
)

state = agent.invoke(
    {"messages": "How to create a Langchain agent?"}
)

state
```

<h3 style="font-size: 20px; font-weight: bold; color: #ff8f1e;">
  11.3 Use connectors with Mistral Conversations & Agents
</h3>

```python
from mistralai.client.models import CustomConnector

response = mistral.beta.conversations.start(
    model="mistral-small-latest",
    inputs="Using your tools, tell me how to create a langchain agent",
    tools=[CustomConnector(connector_id="langchain_documentation")],
)

response.outputs
```

```python
print(response.outputs[0].model_dump_json(indent=2))
```
